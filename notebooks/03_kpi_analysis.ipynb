```python
import pandas as pd
import numpy as np
from pathlib import Path

#read the data
DATA_DIR = Path(r"D:\retail_profitability_project\data\raw\SampleSuperstore.csv")

df = pd.read_csv(DATA_DIR)

print (df.info())

```

```python
#calculate total Revenue
total_revenue = df["Sales"].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
#calculate total profit
total_profit = df["Profit"].sum()
print(f"Total Profit: ${total_profit:,.2f}")
```

```python
# Calculate average profit
average_profit = df["Profit"].mean()
print(f"Average Profit: ${average_profit:,.2f}")
```

```python
#calculate profit margin
profit_margin = total_profit/total_revenue *100
print(f"Profit Margin: {profit_margin:,.2f}%")
```

```python
#calculate average sales
average_revenue = df["Sales"].mean()
print(f"Average Revenue: ${average_revenue:,.2f}")
```

```python
# Calculate total transactions
total_transactions = len(df)
print(f"Total Transactions: {total_transactions}")
```

```python
#calculate total orders made
total_orders = df["Quantity"].sum()
print(f"Total Orders: {total_orders}")
```

```python
#check for customers

print(df["Segment"].value_counts())

```

```python
# Average Determine discount
average_discount = df["Discount"].mean()
print(f"Average Discount: {average_discount:,.2f}%")
```

```python
# Average Order value
aov = total_revenue/ total_orders
print(f"Average Order Value: {aov:,.2f}")
```

```python
kpi_data = {
  "KPI" : ["Total Transactions","Revenue","Profit","Profit Margin","Quantity","Average Order Value","Average Discount","Average Profit","Average Revenue"],
  "value" : [total_transactions,total_revenue,total_profit,profit_margin,total_orders,aov, average_discount,average_profit,average_revenue]
}
table = pd.DataFrame(kpi_data)


def format_kpi(row):
    if row["KPI"] in ["Quantity", "Total Transactions"]:
        return f"{row['value']:,.0f}"
    elif row["KPI"] in ["Profit Margin", "Average Discount", "Average Profit","Average Revenue"]:
        return f"{row['value']:.2f}%"
    else:
        return f"{row['value']:,.2f}"

table["Value"] = table.apply(format_kpi, axis=1)


display(table[["KPI", "Value"]])
```
